```python
import numpy as np
import matplotlib.pyplot as plt
# plt.style.use('seaborn-darkgrid')
plt.rcParams.update({'font.family':'FangSong', 'font.size':12})
# 创建图形
fig, ax = plt.subplots(figsize=(8,6))

# 水平和高度范围
x = np.linspace(-10,10,100)
y = np.linspace(0,12,100)

# 绘制潜热通量区（红色区域）
ax.fill_between(x, 0, 1, color='orange', alpha=0.5, label='高潜热通量区（蒸发强）')

# 绘制低层收敛箭头
for hx in [-8,-6,-4]:
    ax.arrow(hx, 0.5, 1.5, 0, head_width=0.3, head_length=0.4, fc='blue', ec='blue')
for hx in [8,6,4]:
    ax.arrow(hx, 0.5, -1.5, 0, head_width=0.3, head_length=0.4, fc='blue', ec='blue')

# 绘制上升气流（对流塔）
for hy in range(0,10,2):
    ax.arrow(0, hy+1, 0, 1.5, head_width=0.3, head_length=0.4, fc='red', ec='red')

# 绘制高层辐散箭头
for hx in [-3,-2,-1]:
    ax.arrow(hx, 10, -2, 0, head_width=0.3, head_length=0.4, fc='green', ec='green')
for hx in [1,2,3]:
    ax.arrow(hx, 10, 2, 0, head_width=0.3, head_length=0.4, fc='green', ec='green')

# 文字注释
ax.text(-9, 0.2, '低层风汇入', color='blue')
ax.text(-2, 5, '强上升运动（潜热释放）', color='red')
ax.text(-4, 10.5, '高层风扩散（扩线）', color='green')

# 轴设置
ax.set_ylim(0,12)
ax.set_xlim(-10,10)
ax.set_xlabel('水平位置')
ax.set_ylabel('高度 (km)')
ax.set_title('潜热通量增加对大气收敛/扩线的影响示意图')
ax.legend(loc='lower left')

plt.show()
```
